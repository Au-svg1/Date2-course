# 01 — Git, Gitee & GitHub for Reproducible Research

## Course goal

In this course, Git is not used because you are not learning software engineering.

It is used because a research workflow should be:

- reproducible,
- traceable,
- reviewable,
- easy to share.

For classroom use, **Gitee is recommended as the default remote platform**, especially when network access to GitHub is unstable. GitHub is fully acceptable if it works reliably for your group.

By the end of this notebook, you should be able to:

1. understand the difference between Git, Gitee, GitHub and VS Code;
2. create a local Git repository;
3. create a remote repository;
4. connect the local and remote repositories;
5. commit and push your project;
6. use `.gitignore`, `README.md` and `requirements.txt`;
7. record Version 1 → peer review → Version 2;
8. avoid uploading passwords, API keys, or inappropriate datasets.

# 1. Git ≠ Gitee ≠ GitHub ≠ VS Code

These tools are related, but they are not the same thing.

| Tool | What it is | What it does |
|---|---|---|
| Git | Version-control software | Tracks changes on your computer |
| Gitee | Git hosting platform | Stores Git repositories online |
| GitHub | Git hosting platform | Stores Git repositories online |
| VS Code | Code editor | Lets you edit files and use Git through a graphical interface |

A useful mental model is:

```text
Your project folder
      ↓
     Git
      ↓
Local version history
      ↓
Gitee / GitHub
      ↓
Remote shared repository
```

The important concept is:

> **Git tracks changes. Gitee/GitHub host the tracked project online.**

# 3. Configure Your Git Identity

You normally need to do this only once on a computer.

Replace the example values with your own name and email.

```bash
git config --global user.name "Your Name"
git config --global user.email "your_email@example.com"
```

Check your configuration:

```bash
git config --global --list
```

This identity is attached to your Git commits.

It does **not** need to be the same thing as your course nickname or group name.

# 4. Recommended Course Project Structure

For Date2, a repository might look like:

```text
group05-water-quality/
├── README.md
├── requirements.txt
├── .gitignore
├── notebooks/
│   └── analysis.ipynb
├── src/
│   └── build_dataset.py
├── data/
│   └── README.md
└── outputs/
    ├── metadata.csv
    └── data_dictionary.csv
```

Important:

> Do not assume that the complete raw dataset belongs in Git.

Large files, copyrighted files, restricted datasets, passwords and API keys should not be committed simply because they exist in the project folder.

# 5. Create a `.gitignore` Before Your First Commit

A `.gitignore` tells Git which files should normally be ignored.

For this course, a reasonable starting point is:

```text
.ipynb_checkpoints/
__pycache__/
*.pyc
.DS_Store
.env
.vscode/
```

Depending on the project, you may also ignore large local data folders.

For example:

```text
data/raw/
outputs/temp/
```

But do not blindly ignore important outputs required for reproducibility.

The decision should be documented in the README.

# 6. Create a README

Every repository should have a `README.md`.

A minimal course README should explain:

```markdown
# Project title

## Team
Group X

## Research / data question
...

## Input
...

## Output
...

## How to run
...

## Required packages
...

## Data source
...

## Known limitations
...

## AI Assistance
...
```

A good test is:

> Can another group understand the project within five minutes without asking you questions?

# 7. What Is `requirements.txt`?

`requirements.txt` records the Python packages required by the project.

Example:

```text
pandas
numpy
matplotlib
pdfplumber
plotly
scikit-learn
```

Another user can install them with:

```bash
python -m pip install -r requirements.txt
```

For this course, the purpose is reproducibility—not package perfection.

Do not add packages that your project does not actually use.

# 8. Start Git Tracking in Your Existing Project

Open a terminal in the root folder of your project.

Check the current folder:

```bash
pwd
```

On Windows PowerShell you may use:

```powershell
Get-Location
```

Then initialize Git:

```bash
git init -b main
```

If your installed Git version does not support `-b main`, use:

```bash
git init
git branch -M main
```

Check the project:

```bash
git status
```

You should now see untracked files.

# 9. First Commit

Before adding files, inspect the folder and make sure it contains no:

- passwords,
- API keys,
- access tokens,
- personal credentials,
- private data,
- inappropriate large files.

Then:

```bash
git add .
```

Check what will be committed:

```bash
git status
```

If everything is appropriate:

```bash
git commit -m "Initial course project"
```

Think of a commit as a named snapshot of your project.

# 10. Create the Remote Repository — Gitee Recommended for Class

For classroom use, the simplest workflow is:

1. Sign in to Gitee.
2. Create a new repository.
3. Give it a clear name, for example:

```text
group05-water-quality
```

4. Choose the repository visibility appropriate for the course.
5. For an existing local project, it is usually easiest to create an **empty remote repository** rather than adding another README to the remote first.
6. Copy the repository's HTTPS URL.

It will look conceptually like:

```text
https://gitee.com/YOUR_USERNAME/YOUR_REPOSITORY.git
```

Do not paste your password, token, or other credentials into the notebook.

# 11. Connect Local Git to Gitee

Replace the example URL with the URL shown by your own Gitee repository.

```bash
git remote add origin https://gitee.com/YOUR_USERNAME/YOUR_REPOSITORY.git
```

Check it:

```bash
git remote -v
```

Then push:

```bash
git push -u origin main
```

After the first successful push, later updates are usually just:

```bash
git push
```

If your remote uses a different default branch name, check the repository page before changing commands blindly.

# 12. GitHub Alternative

If GitHub works reliably for you, the workflow is almost identical.

Create an empty GitHub repository and copy its HTTPS URL.

Then:

```bash
git remote add origin https://github.com/YOUR_USERNAME/YOUR_REPOSITORY.git
git remote -v
git push -u origin main
```

Important:

GitHub authentication may require a browser sign-in, credential manager, SSH key, or token-based authentication depending on your setup.

Do **not** put a password or access token inside:

- a notebook,
- Python script,
- README,
- `.env` file that is committed,
- or a screenshot shared publicly.

# 13. The Minimum Git Workflow You Need for This Course

After the repository is set up, most classroom work is only:

```bash
git status
git add .
git commit -m "Describe what changed"
git push
```

Conceptually:

```text
Edit files
   ↓
git status
   ↓
git add .
   ↓
git commit
   ↓
git push
```

Use informative commit messages.

Better:

```text
Add availability analysis
Fix metadata field definitions
Revise dataset after peer review
```

Less useful:

```text
update
test
aaa
finalfinal2
```

# 14. Version 1 → Peer Review → Version 2

This is where Git becomes useful for Date2.

Your group competition can use the following workflow:

```text
Build data product
      ↓
Commit Version 1
      ↓
Another group reviews it
      ↓
Prepare response_to_reviewers.md
      ↓
Revise the project
      ↓
Commit Version 2
      ↓
Prepare mock data release
```

Example commit sequence:

```bash
git add .
git commit -m "Submit data product Version 1"
git push
```

After peer review:

```bash
git add .
git commit -m "Revise dataset after peer review"
git push
```

Now the repository contains evidence of how the research product changed.

# 15. Optional: Create Simple Version Tags

Tags are optional for this course, but useful if you want a clear Version 1 and Version 2.

After the Version 1 commit:

```bash
git tag v1.0-classroom
git push origin v1.0-classroom
```

After revision:

```bash
git tag v2.0-classroom
git push origin v2.0-classroom
```

Do not use this if your group is still struggling with basic Git.

A clear commit history is sufficient.

# 16. Using Git Through VS Code

VS Code can perform most basic Git operations without typing commands.

### Typical workflow

1. Open the project folder in VS Code.
2. Click the **Source Control** icon in the left sidebar.
3. Modified/untracked files appear under Changes.
4. Review the changed files.
5. Stage the files you want to include.
6. Enter a commit message.
7. Commit.
8. Sync / Push.

Even if you use the graphical interface, you should still understand:

```text
Stage = git add
Commit = git commit
Push = git push
Pull = download remote changes
```

If something behaves unexpectedly, open the integrated terminal and run:

```bash
git status
```

This is often the fastest way to understand the current repository state.

# 17. Do Not Upload Everything

Before every push, ask:

### Should this file be public or shared?

Do not commit:

- passwords,
- API keys,
- access tokens,
- private credentials,
- confidential research data,
- personal information,
- files that you do not have permission to redistribute,
- unnecessary huge files.

For this course, the complete CNEMC-derived classroom dataset does **not** need to be pushed to Gitee/GitHub.

The repository can contain:

- code,
- notebook,
- README,
- requirements,
- metadata structure,
- data dictionary,
- small permitted example outputs,
- documentation.

The full classroom data can remain local when appropriate.

# 18. Common Error — `remote origin already exists`

If you run:

```bash
git remote add origin ...
```

and see:

```text
error: remote origin already exists
```

first inspect the existing remote:

```bash
git remote -v
```

If it is wrong, change it:

```bash
git remote set-url origin YOUR_CORRECT_URL
```

Do not keep adding multiple `origin` remotes.

# 19. Common Error — Push Rejected

A push may be rejected because the remote repository already contains commits that your local repository does not have.

This often happens when you created a README online and also created a different README locally.

For beginners, the easiest prevention is:

> When connecting an existing local project, create an empty remote repository.

If you already have different histories, do not randomly use `--force`.

Ask the instructor or inspect:

```bash
git status
git log --oneline --all --graph
git remote -v
```

before changing history.

# 20. Common Error — Authentication Failed

If authentication fails:

1. Confirm that the remote URL is correct.
2. Confirm that you are signed in to the correct Gitee/GitHub account.
3. Follow the platform's current authentication method.
4. If using VS Code, you may be asked to authenticate in your browser.
5. Never paste access credentials into course notebooks.

Check the remote:

```bash
git remote -v
```

The repository URL is safe to show to the instructor.

Your password/token is not.

# 21. Common Error — Wrong Folder

Students sometimes run Git commands one directory above or below the actual project.

Check:

```bash
git status
```

and:

```bash
pwd
```

The project root should contain files such as:

```text
README.md
requirements.txt
notebooks/
src/
```

Do not run `git init` repeatedly in nested folders unless you intentionally want nested repositories.

## Five-minute peer test

Give your repository link to another team.

Without explaining anything verbally, ask them:

1. What does this project do?
2. What are the inputs?
3. What are the outputs?
4. How should the analysis be run?
5. What is the biggest documented limitation?

If they cannot answer these questions, improve your README.

# 23. Suggested `AI Assistance` Section

Add a section like this to your README:

```markdown
## AI Assistance

AI tools used:
- ChatGPT
- Codex

Tasks assisted by AI:
- debugging
- code refactoring
- README drafting

Human verification:
- checked indicator definitions against source material
- verified missing-value handling
- checked output row counts
- reviewed metadata fields

Decisions not delegated to AI:
- inclusion criteria
- final interpretation
- release decision
```

The goal is transparency—not banning AI.

# 24. Quick Command Reference

```bash
# Check Git
git --version

# Check repository state
git status

# See remote repository
git remote -v

# Stage files
git add .

# Commit
git commit -m "Describe the change"

# Push
git push

# Download new remote changes
git pull

# See compact history
git log --oneline

# Change remote URL
git remote set-url origin YOUR_URL
```

For this course, you do not need advanced commands such as rebase unless you already know how to use them.

# 25. Final Classroom Workflow

The recommended Date2 workflow is:

```text
Local project
    ↓
Git repository
    ↓
Version 1 commit
    ↓
Push to Gitee / GitHub
    ↓
Peer review
    ↓
Response to reviewers
    ↓
Revision
    ↓
Version 2 commit
    ↓
Final Git push
    ↓
Mock research-data release
```

### Remember

> Git is not just a way to upload files.

It provides a traceable history of how your research product changed.

For this course:

> **A simple repository that another group can understand and reproduce is better than a complicated repository that nobody can run.**
