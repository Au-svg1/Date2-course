## 1. Basic Environment Diagnostic

Run the following cell first.

If you ask the instructor for help, please provide the output of this cell.

````python
import sys
import os
import platform
from pathlib import Path

print("=" * 70)
print("PYTHON ENVIRONMENT DIAGNOSTIC")
print("=" * 70)

print("\nOperating system:")
print(platform.platform())

print("\nPython version:")
print(sys.version)

print("\nPython executable:")
print(sys.executable)

print("\nWorking directory:")
print(Path.cwd())

print("\nConda environment:")
print(os.environ.get("CONDA_DEFAULT_ENV", "Not detected"))

print("\nVirtual environment:")
print(os.environ.get("VIRTUAL_ENV", "Not detected"))

print("\n" + "=" * 70)
````

### How to read the result

The most important line is:

```text

...
```

Examples:

```text

```

or

```text

```

```text

```

This path tells you **which Python is actually running this notebook**.

If the notebook works and the required packages are available, you do not need to change environments just because the path looks different from the instructor's computer.

## 2. Check Required Course Packages

This cell checks whether the main packages used in Date1–Date3 are available.

````python
packages = {
    "numpy": "NumPy",
    "pandas": "pandas",
    "matplotlib": "Matplotlib",
    "requests": "requests",
    "bs4": "BeautifulSoup4",
    "openpyxl": "openpyxl",
    "pdfplumber": "pdfplumber",
    "sklearn": "scikit-learn",
    "plotly": "Plotly",
}

missing = []

for module, display_name in packages.items():
    try:
        __import__(module)
        print(f"✓ {display_name}")
    except ImportError:
        print(f"✗ {display_name}  -- not installed in THIS environment")
        missing.append(display_name)

print("\n" + "-" * 60)

if missing:
    print("Missing packages:")
    for pkg in missing:
        print(" -", pkg)
else:
    print("All core course packages are available.")
````

## 3. If a Package Is Missing

Inside Jupyter, the safest method is usually:

```python
%pip install package_name
```

For example:

```python
%pip install pdfplumber
```

Why use `%pip`?

Because `%pip` installs into the Python environment connected to the **current Jupyter kernel**.

This avoids a very common problem:

> `pip install` succeeds in one Python environment, but the notebook is using another Python environment.

Only install packages that are actually missing.

# 11. Common Problem 4 — Wrong Working Directory

A correct Python environment does not guarantee that Python can find your files.

Check:

````python
from pathlib import Path
print("Current working directory:")
print(Path.cwd())
````

For this course, prefer relative paths.

Good:

```python
from pathlib import Path

PROJECT_DIR = Path.cwd()
DATA_DIR = PROJECT_DIR / "data"
```

Avoid machine-specific paths when possible:

```python
"/Users/yourname/Desktop/course/data/file.csv"
```

A path that works on one student's computer may fail on everyone else's computer.

# 12. Quick Environment Health Check

Run this cell when you want a compact summary.

````python
import sys
import os
from pathlib import Path
import importlib

required = [
    "numpy",
    "pandas",
    "matplotlib",
    "requests",
    "bs4",
    "openpyxl",
    "pdfplumber",
    "sklearn",
    "plotly",
]

print("Python:", sys.version.split()[0])
print("Executable:", sys.executable)
print("Environment:", os.environ.get("CONDA_DEFAULT_ENV", "Not detected"))
print("Folder:", Path.cwd())
print("\nPackages:")

all_ok = True

for pkg in required:
    try:
        importlib.import_module(pkg)
        print(f"  ✓ {pkg}")
    except ImportError:
        print(f"  ✗ {pkg}")
        all_ok = False

print("\nStatus:")
if all_ok:
    print("✓ READY FOR THE COURSE")
else:
    print("⚠ Some packages are missing. Install only the missing packages.")
````
